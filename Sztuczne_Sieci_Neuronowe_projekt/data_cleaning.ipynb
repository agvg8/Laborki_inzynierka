```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df_kostki = pd.read_csv("C:\\Users\\Aga\\PycharmProjects\\SSN_Projekt\\dice_ext.csv", header=None)
```

```python
# Sprawdzenie liczby wierszy
num_rows = df_kostki.shape[0]
print(f'Liczba wierszy w pliku: {num_rows}')
```

```python
# Sprawdzenie poprawności formatu danych
# Sprawdzamy, czy pierwsze dwie kolumny mają odpowiednie wartości
valid_labels = df_kostki.iloc[:, 0].isin([0, 1])  # Sprawdzenie, czy etykiety są 0 lub 1
valid_dice_values = df_kostki.iloc[:, 1].between(0, 6) | df_kostki.iloc[:, 1].between(7, 10)  # 1-6 to liczba oczek, 7-10 to typy niepoprawności
valid_data_format = valid_labels & valid_dice_values & (df_kostki.iloc[:, 2:].apply(lambda x: len(x) == 784, axis=1))

print(f'Liczba poprawnych wierszy: {valid_data_format.sum()}')
```

```python
# Wydrukowanie obrazu z wiersza o indeksie (numer studenta % 70000)
student_index = 120627  # Przykładowy numer indeksu
index_to_print = student_index % 70000
```

```python
# Wydruk obrazu
image_data = df_kostki.iloc[index_to_print, 2:].values.reshape(28, 28)
plt.imshow(image_data, cmap='gray')
plt.title(f'Obraz z indeksem {index_to_print}')
plt.show()
```
